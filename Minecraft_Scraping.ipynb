```python
from google_play_scraper import reviews_all, Sort # Mengimpor pustaka google_play_scraper untuk mengakses ulasan dan informasi aplikasi dari Google Play Store.
import pandas as pd
```

# Scapping Dataset

Pada proyek ini, kita akan melakukan analisis sentimen terhadap ulasan dan pendapat pengguna terkait game "Minecraft" di Play Store. Analisis sentimen akan membantu kita untuk memahami bagaimana pengguna merasakan dan mengungkapkan pandangan mereka terhadap game ini.

Dengan menggunakan berbagai teknik pemrosesan teks dan algoritma machine learning, kita akan mencoba mengidentifikasi apakah ulasan pengguna terhadap game "Minecraft" cenderung positif, negatif, atau netral. Hasil analisis sentimen ini dapat memberikan wawasan berharga kepada pengembang game untuk meningkatkan pengalaman pengguna. Mari kita mulai menjelajahi ulasan pengguna dan menganalisis sentimen di sekitar game "Minecraft" dengan lebih mendalam.

```python
# Mengambil semua ulasan dari aplikasi dengan ID 'com.mojang.minecraftpe' di Google Play Store.
# Proses scraping mungkin memerlukan beberapa saat tergantung pada jumlah ulasan yang ada.
scrapreview = reviews_all(
    'com.mojang.minecraftpe',     # ID aplikasi
    lang='en',                    # Bahasa ulasan (default: 'en')
    country='us',                 # Negara (default: 'us')
    sort=Sort.MOST_RELEVANT,      # Urutan ulasan (default: Sort.MOST_RELEVANT)
    count=1000000                 # Jumlah maksimum ulasan yang ingin diambil
)
```

# Loading Dataset

```python
app_reviews_df = pd.DataFrame(scrapreview)
app_reviews_df.shape
```

```python
app_reviews_df.to_csv('Dataset/Minecraft_Review.csv', index=False)
app_reviews_df
```
